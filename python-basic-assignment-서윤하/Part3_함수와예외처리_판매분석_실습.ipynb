```python
import csv

with open("자동차_판매_데이터.csv", "r", encoding = 'utf-8-sig') as file:
    reader = csv.DictReader(file)
    sales_data = list(reader)
```

#### 3-1. 판매금액 구간 분류 함수 만들기

```python
def function(amount,sale_id):
    if type(amount) != list or type(sale_id) != list:
        raise TypeError("입력값이 잘못되었습니다.")
    if len(amount) != len(sale_id):
        raise ValueError("두 리스트의 길이가 다릅니다.")
    if len(amount)==0 or len(sale_id)==0:
        raise ValueError("값을 입력하세요.")

    result = {
        "고가": [],
        "중가": [],
        "저가": []
    }

    for price, id in zip(amount, sale_id):
        if price >= 70000000:
            result["고가"].append(id)

        elif price >= 40000000:
            result['중가'].append(id)

        else:
            result["저가"].append(id)

    return result

print(function([30000000, 45000000, 80000000], ['1234ad', '2345ef', '3456cd']))
```

#### 3-2. 배송 확인 거래 검색 함수 만들기

```python
def find_delivery_check_sales(sale_ids, delivery_days, stock_statuses):
    if type(sale_ids) != list or type(delivery_days) != list or type(stock_statuses)!= list:
        raise TypeError("입력값이 잘못되었습니다.")
    if len(sale_ids) != len(delivery_days) != len(stock_statuses):
        raise ValueError("리스트의 길이가 다릅니다.")

    result = []

    for ids, day, status in zip(sale_ids, delivery_days, stock_statuses):
        if day >= 20 and status != "출고완료":
            result.append(ids)

    return result

print(find_delivery_check_sales(['S2025010207', 'S2025010498'], [21, 18], ['재고보유', '출고완료']))
# print(find_delivery_check_sales({'S2025010207', 'S2025010498'}, [21, 18], ['재고보유', '출고완료']))
# print(find_delivery_check_sales(['S2025010207', 'S2025010498'], [18], ['재고보유', '출고완료']))
```

#### 3-3. 배송 점검 결과를 CSV로 저장하기

```python
selected_sales = []

for row in sales_data:
    selected_id = row['SaleID']
    selected_days = int(row['DeliveryDays'])
    selected_status = row['StockStatus']

    if find_delivery_check_sales([selected_id], [selected_days], [selected_status]):
        selected_sales.append(row)

fieldnames = reader.fieldnames
with open("delivery_check_sales.csv", "w", encoding='utf-8-sig', newline="") as file:
    writer = csv.DictWriter(file, fieldnames=fieldnames)
    writer.writeheader()
    writer.writerows(selected_sales)

with open("delivery_check_sales.csv", "r", encoding='utf-8-sig') as file:
    reader = csv.DictReader(file)
    data = list(reader)

print('저장 전 거래 수:', len(selected_sales))
print('저장 후 거래 수:', len(data))
```
