```python
import csv

with open("자동차_판매_데이터.csv", "r", encoding = 'utf-8-sig') as file:
    reader = csv.DictReader(file)
    sales_data = list(reader)
```

```python
high_count = 0
middle_count = 0
low_count = 0

high_price = []
middle_price = []
low_price = []

for price in sales_data:
    amount = int(price["FinalSaleAmount"])
    sale_id = price["SaleID"]
    
    if amount >= 70000000:
        high_count += 1
        high_price.append(sale_id)

    elif amount >= 40000000:
        middle_count += 1
        middle_price.append(sale_id)

    else:
        low_count += 1
        low_price.append(sale_id)

print('고가 거래 수:', high_count)
print('고가 거래 ID 5개:',high_price[0:5])
print()
print('중가 거래 수:', middle_count)
print('중가 거래 ID 5개:', middle_price[0:5])
print()
print('저가 거래 수:', low_count)
print('저가 거래 ID 5개:', low_price[0:5])
print()
print('총 거래 수:', len(sales_data))
```

#### 2-2. (심화) 배송 확인이 필요한 거래 찾기

```python
delivery_check = 0
delivery_check_id = []

for delivery in sales_data:
    delivery_days = int(delivery["DeliveryDays"])
    stock_status = delivery["StockStatus"]
    sale_id = delivery["SaleID"]

    if delivery_days >= 20 and stock_status != "출고완료":
        delivery_check += 1
        delivery_check_id.append(sale_id)

print("배송 확인이 필요한 거래:", delivery_check)
print("배송 확인이 필요한 거래 ID 5개:", delivery_check_id[0:5])
```

#### 2-3. (심화) 고액 반품 거래 찾기

```python
return_total_amount = 0
return_id = []

for returns in sales_data:
    sale_amount = int(returns["FinalSaleAmount"])
    return_status = returns["IsReturned"]
    sale_id = returns["SaleID"]

    if sale_amount >= 70000000 and return_status == "Y":
        return_total_amount += sale_amount
        return_id.append(sale_id)

avg_return_amount = return_total_amount / len(return_id)

print("고액 반품 거래 총 금액:", return_total_amount)
print("고액 반품 거래 평균 금액:", avg_return_amount)
print("고액 반품 거래 수:", len(return_id))
print("고액 반품 거래 ID:",return_id)
```
